```python
import pandas as pd
import numpy as np
import plotly.express as px
from datasist.structdata import detect_outliers
```

```python
# loading the data 

df = pd.read_csv("/kaggle/input/datasets/maria561/europe-electrification-data-energy-ember/europe_electrification.csv",
                 index_col=False)

df.info()
```

```python
# making features headers in lower case for better readability

df.columns=df.columns.str.lower()
```

```python
# checking variance in categorical features 

for col in df.select_dtypes(include = 'object'):
    print(f"no of unique: {df[col].nunique()}")
    print(df[col].value_counts().sort_values(ascending=True))
    print('*'*50)
```

```python
# dropping redundant features

df = df.drop(["country_code","parent_sector_name"],axis = 1)
```

```python
# checking duplicates

duplicates = df.duplicated().sum()
duplicates
```

```python
df.describe(include="number")
```

```python
# exporting clean dataset

import os

output_path = "/kaggle/working/europe_electrification_.csv"

df.to_csv(output_path, index=False)
```
